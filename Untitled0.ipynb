```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

# Locate the Mathematics CSV file
csv_path = list(
    extract_path.rglob("student-mat.csv")
)[0]

# Load the dataset
df = pd.read_csv(csv_path, sep=';')

print("Dataset loaded successfully!")
```

```python

# Display first five records
print(df.head())
```

```python

# Number of rows and columns
print("Dataset Shape:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())
```

```python

# Select numerical columns
columns = [
    'age', 'studytime', 'absences',
    'G1', 'G2', 'G3'
]

# Calculate descriptive statistics
print(df[columns].describe())

# Calculate mode
print("\nMode:")
print(df[columns].mode().iloc[0])

```

```python

# Select columns for outlier detection
outlier_columns = ['G1', 'G2', 'G3', 'absences']

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print("\nColumn:", col)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import zipfile
from google.colab import files

# Create a folder for graphs
import os
os.makedirs('/content/EDA_Graphs', exist_ok=True)

# 1. Box Plot
plt.figure(figsize=(8, 5))
sns.boxplot(y=df['G3'])
plt.title('Box Plot of Final Grades')
plt.ylabel('Final Grade (G3)')
plt.savefig('/content/EDA_Graphs/boxplot.png',
            dpi=300, bbox_inches='tight')
plt.show()
plt.close()

# 2. Scatter Plot: Study Time vs Final Grade
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='studytime', y='G3')
plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')
plt.savefig('/content/EDA_Graphs/scatter_studytime.png',
            dpi=300, bbox_inches='tight')
plt.show()
plt.close()

# 3. Scatter Plot: G2 vs G3
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='G2', y='G3')
plt.title('Second Period Grade vs Final Grade')
plt.xlabel('Second Period Grade (G2)')
plt.ylabel('Final Grade (G3)')
plt.savefig('/content/EDA_Graphs/scatter_g2_g3.png',
            dpi=300, bbox_inches='tight')
plt.show()
plt.close()

# 4. Correlation Heatmap
plt.figure(figsize=(10, 7))

correlation = df[
    ['age', 'studytime', 'absences', 'G1', 'G2', 'G3']
].corr()

sns.heatmap(correlation, annot=True,
            cmap='coolwarm', fmt='.2f')

plt.title('Student Performance Correlation Heatmap')
plt.savefig('/content/EDA_Graphs/correlation_heatmap.png',
            dpi=300, bbox_inches='tight')
plt.show()
plt.close()

# Create ZIP file containing all graphs
zip_path = '/content/EDA_Graphs.zip'

with zipfile.ZipFile(zip_path, 'w') as zip_file:
    for filename in os.listdir('/content/EDA_Graphs'):
        filepath = os.path.join('/content/EDA_Graphs', filename)
        zip_file.write(filepath, arcname=filename)

print("All four graphs generated successfully!")

# Download all graphs together
files.download(zip_path)
```

```python

!pip install ipywidgets -q

import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

print("Interactive simulation is ready!")
```

```python

# Interactive Student Performance EDA

grade_choice = widgets.Dropdown(
    options=['G1', 'G2', 'G3'],
    value='G3',
    description='Grade:'
)

graph_choice = widgets.Dropdown(
    options=[
        'Histogram',
        'Box Plot',
        'Scatter Plot',
        'Correlation Heatmap',
        'Descriptive Statistics'
    ],
    description='Analysis:'
)

variable_choice = widgets.Dropdown(
    options=['studytime', 'age', 'absences', 'G1', 'G2'],
    value='G2',
    description='Compare:'
)

output = widgets.Output()

def run_simulation(grade, graph, variable):

    with output:
        clear_output(wait=True)

        if graph == 'Histogram':
            plt.figure(figsize=(8, 5))
            sns.histplot(df[grade], bins=10, kde=True)
            plt.title(f'Distribution of {grade}')
            plt.xlabel(grade)
            plt.ylabel('Number of Students')
            plt.show()

        elif graph == 'Box Plot':
            plt.figure(figsize=(8, 5))
            sns.boxplot(y=df[grade])
            plt.title(f'Box Plot of {grade}')
            plt.show()

        elif graph == 'Scatter Plot':
            plt.figure(figsize=(8, 5))
            sns.scatterplot(x=df[variable], y=df[grade])
            plt.title(f'{variable} vs {grade}')
            plt.xlabel(variable)
            plt.ylabel(grade)
            plt.show()

        elif graph == 'Correlation Heatmap':
            plt.figure(figsize=(9, 6))
            corr = df.select_dtypes(
                include='number'
            ).corr()
            sns.heatmap(
                corr,
                annot=True,
                cmap='coolwarm',
                fmt='.2f'
            )
            plt.title('Student Performance Correlation')
            plt.show()

        elif graph == 'Descriptive Statistics':
            print(df[grade].describe())
            print("\nMean:", df[grade].mean())
            print("Median:", df[grade].median())
            print("Mode:", df[grade].mode()[0])
            print("Standard Deviation:", df[grade].std())

        print("\nSelected Grade:", grade)
        print("Selected Analysis:", graph)

interactive = widgets.interactive_output(
    run_simulation,
    {
        'grade': grade_choice,
        'graph': graph_choice,
        'variable': variable_choice
    }
)

display(widgets.VBox([
    grade_choice,
    graph_choice,
    variable_choice,
    output
]))

run_simulation('G3', 'Histogram', 'G2')
```
